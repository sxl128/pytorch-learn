```python
import torch

points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
second_points = points[1]
second_points.storage_offset()
```

```python
second_points.size()
```

```python
# stride[i] 表示在内存上 (storage) 沿第 i 维走 1 步要跳过多少元素
points.stride()

# accessing an element i, j in a 2D tensor results in accessing
# the storage_offset + stride[0] * i + stride[1] * j element in the storage
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points
```

```python
points_t = points.t()  # shorthand alternative to transpose for two-dimensional tensors
points_t
```

```python
id(points.untyped_storage()) == id(points_t.untyped_storage())
```

```python
points.stride()
```

```python
points_t.stride()
```

```python
# transposing in higher dimensions 
some_t = torch.ones(3, 4, 5)
transpose_t = some_t.transpose(0, 2)
some_t.shape
```

```python
transpose_t.shape
```

```python
some_t.stride()
```

```python
transpose_t.stride()
```

```python
# Contiguous tensors
# Some tensor operations in pytorch only work on contiguous tensors
points.is_contiguous()
```

```python
points_t.is_contiguous()
```

```python
# we can obtain a new contiguous tensor from noncontiguous one using contiguous method
# the content of the tensor will be the same, but the stride will change
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points_t = points.t()
points_t
```

```python
points_t.storage()
```

```python

points_t.stride()
```

```python
points_t_cont = points_t.contiguous()
points_t_cont
```

```python
points_t_cont.stride()
```

```python
# contiguous will make a new reshuffled storage
points_t_cont.storage()
```

```python
points_t.storage()
```
